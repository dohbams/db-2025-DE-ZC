```python
import pandas as pd
```

```python
df = pd.read_parquet("yellow_tripdata_2024-01.parquet")
```

```python
df.head(10)
```

```python
from sqlalchemy import create_engine
```

```python
engine =  create_engine('postgresql://root:root@localhost:5432/ny_taxi_db')
```

```python
engine.connect()
```

```python
schema = pd.io.sql.get_schema(df, "yellow_taxi_data", con= engine)
```

```python
print(schema)
```

```python
# Upload the full DataFrame at once
%time df.to_sql("yellow_taxi_data", con=engine, if_exists="append", index=False)
```

```python
lookup = pd.read_csv("taxi_zone_lookup.csv")
```

```python
lookup_schema = pd.io.sql.get_schema(lookup, "zones", con= engine)
```

```python
print(lookup_schema)
```

```python
# Upload the full DataFrame at once
%time lookup.to_sql("zones", con=engine, if_exists="replace", index=False)
```
